# Sales Analysis
Independent exploratory calculations from the cleaned dataset.

```python
import pandas as pd
df = pd.read_csv('../data/processed/cleaned_superstore.csv')
print('Revenue:', df['Sales'].sum())
print('Profit:', df['Profit'].sum())
print('Orders:', df['Order ID'].nunique())
```

```python
df.groupby('Category')[['Sales','Profit']].sum().sort_values('Sales', ascending=False)
```

```python
df.groupby('Region')[['Sales','Profit']].sum().sort_values('Sales', ascending=False)
```

```python
df.groupby('Product Name')['Profit'].sum().nsmallest(10)
```
